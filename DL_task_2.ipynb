**Task 2** — Xavier, Kaiming and Gradient Vanishing

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

torch.manual_seed(42)

# Xavier initialization
def xavier_init(layer):
    torch.nn.init.xavier_uniform_(layer.weight)
    torch.nn.init.zeros_(layer.bias)

# Kaiming initialization
def kaiming_init(layer):
    torch.nn.init.kaiming_uniform_(layer.weight, nonlinearity="relu")
    torch.nn.init.zeros_(layer.bias)

# Create 20-layer network
layers = []

for i in range(20):
    layers.append(torch.nn.Linear(50, 50))

# Xavier initialization
for layer in layers:
    xavier_init(layer)

x = torch.randn(32, 50)

activation_means = []
activation_vars = []

for layer in layers:
    x = torch.relu(layer(x))

    activation_means.append(x.mean().item())
    activation_vars.append(x.var().item())

plt.plot(activation_means)
plt.title("Mean of Activations")
plt.xlabel("Layer")
plt.ylabel("Mean")
plt.show()

plt.plot(activation_vars)
plt.title("Variance of Activations")
plt.xlabel("Layer")
plt.ylabel("Variance")
plt.show()
```
