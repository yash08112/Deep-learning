**Task 5** — Batch Normalization and Layer Normalization

```python
import torch

x = torch.randn(4, 3)

# -----------------
# Batch Normalization
# -----------------

mean = x.mean(dim=0, keepdim=True)
variance = x.var(dim=0, unbiased=False, keepdim=True)

x_batch_norm = (x - mean) / torch.sqrt(variance + 1e-5)

print("Batch Normalization:")
print(x_batch_norm)


# -----------------
# Layer Normalization
# -----------------

mean = x.mean(dim=1, keepdim=True)
variance = x.var(dim=1, unbiased=False, keepdim=True)

x_layer_norm = (x - mean) / torch.sqrt(variance + 1e-5)

print("\nLayer Normalization:")
print(x_layer_norm)
```
