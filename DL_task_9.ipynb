**Task 9** — Anchor-Free Object Detection Loss

```python
import torch


def ciou_loss(box1, box2):

    # box format:
    # x1, y1, x2, y2

    x1 = torch.max(box1[0], box2[0])
    y1 = torch.max(box1[1], box2[1])

    x2 = torch.min(box1[2], box2[2])
    y2 = torch.min(box1[3], box2[3])

    intersection = torch.clamp(
        x2 - x1, min=0
    ) * torch.clamp(
        y2 - y1, min=0
    )

    area1 = (
        (box1[2] - box1[0]) *
        (box1[3] - box1[1])
    )

    area2 = (
        (box2[2] - box2[0]) *
        (box2[3] - box2[1])
    )

    union = area1 + area2 - intersection

    iou = intersection / (union + 1e-7)

    return 1 - iou


pred_box = torch.tensor([
    10., 10., 50., 50.
])

target_box = torch.tensor([
    12., 12., 48., 48.
])

loss = ciou_loss(pred_box, target_box)

print("Box Loss:", loss.item())


# Classification loss
classification_loss = torch.nn.functional.cross_entropy(
    torch.tensor([[2.0, 1.0, 0.5]]),
    torch.tensor([0])
)

print("Classification Loss:",
      classification_loss.item())

total_loss = loss + classification_loss

print("Total Loss:", total_loss.item())
```
