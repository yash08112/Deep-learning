**Task 10** — WGAN-GP

```python
import torch
import torch.nn as nn


# Generator
class Generator(nn.Module):

    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(
            nn.Linear(100, 128),
            nn.ReLU(),

            nn.Linear(128, 256),
            nn.ReLU(),

            nn.Linear(256, 784),
            nn.Tanh()
        )

    def forward(self, z):
        return self.model(z)


# Critic
class Critic(nn.Module):

    def __init__(self):

        super().__init__()

        self.model = nn.Sequential(
            nn.Linear(784, 256),
            nn.LeakyReLU(0.2),

            nn.Linear(256, 128),
            nn.LeakyReLU(0.2),

            nn.Linear(128, 1)
        )

    def forward(self, x):
        return self.model(x)


G = Generator()
C = Critic()

batch_size = 16

z = torch.randn(batch_size, 100)

fake_images = G(z)

print("Generated images:",
      fake_images.shape)


# Gradient penalty
real_images = torch.randn(batch_size, 784)

alpha = torch.rand(batch_size, 1)

interpolated = (
    alpha * real_images +
    (1 - alpha) * fake_images
)

interpolated.requires_grad_(True)

critic_output = C(interpolated)

gradients = torch.autograd.grad(
    outputs=critic_output,
    inputs=interpolated,
    grad_outputs=torch.ones_like(critic_output),
    create_graph=True
)[0]

gradient_norm = gradients.view(
    batch_size, -1
).norm(2, dim=1)

gradient_penalty = (
    (gradient_norm - 1) ** 2
).mean()

print("Gradient Penalty:",
      gradient_penalty.item())
```
