**Task 4** — Momentum, RMSprop and Adam from Scratch

```python
import torch

# Simple parameter
w = torch.tensor(5.0)
target = torch.tensor(0.0)

lr = 0.01

# -----------------
# Momentum
# -----------------

velocity = 0

for i in range(10):

    gradient = 2 * (w - target)

    velocity = 0.9 * velocity + gradient

    w = w - lr * velocity

print("Momentum result:", w.item())


# -----------------
# RMSprop
# -----------------

w = torch.tensor(5.0)
square_avg = 0

for i in range(10):

    gradient = 2 * (w - target)

    square_avg = 0.9 * square_avg + 0.1 * gradient ** 2

    w = w - lr * gradient / (torch.sqrt(square_avg) + 1e-8)

print("RMSprop result:", w.item())


# -----------------
# Adam
# -----------------

w = torch.tensor(5.0)

m = 0
v = 0

beta1 = 0.9
beta2 = 0.999

for t in range(1, 11):

    gradient = 2 * (w - target)

    m = beta1 * m + (1 - beta1) * gradient
    v = beta2 * v + (1 - beta2) * gradient ** 2

    m_hat = m / (1 - beta1 ** t)
    v_hat = v / (1 - beta2 ** t)

    w = w - lr * m_hat / (torch.sqrt(v_hat) + 1e-8)

print("Adam result:", w.item())
```
