**Task 13** — Temporal Convolutional Network

```python
import torch
import torch.nn as nn


class TCNBlock(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        dilation
    ):

        super().__init__()

        padding = (3 - 1) * dilation

        self.conv = nn.Conv1d(
            in_channels,
            out_channels,
            kernel_size=3,
            padding=padding,
            dilation=dilation
        )

        self.relu = nn.ReLU()

    def forward(self, x):

        x = self.conv(x)

        # Remove future information
        x = x[:, :, :-self.conv.padding[0]]

        return self.relu(x)


class TCN(nn.Module):

    def __init__(self):

        super().__init__()

        self.block1 = TCNBlock(
            1, 16, dilation=1
        )

        self.block2 = TCNBlock(
            16, 32, dilation=2
        )

        self.block3 = TCNBlock(
            32, 16, dilation=4
        )

        self.output = nn.Conv1d(
            16, 1, 1
        )

    def forward(self, x):

        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)

        return self.output(x)


model = TCN()
x = torch.randn(4, 1, 100)
output = model(x)
print("Input shape:", x.shape)
print("Output shape:", output.shape)
```
