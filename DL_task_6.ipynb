**Task 6** — Bidirectional LSTM from Scratch

```python
import torch

torch.manual_seed(42)

input_size = 4
hidden_size = 5
sequence_length = 6

# Input sequence
x = torch.randn(sequence_length, input_size)

# Parameters
W = torch.randn(input_size + hidden_size, 4 * hidden_size)
b = torch.zeros(4 * hidden_size)


def lstm_forward(sequence):

    h = torch.zeros(hidden_size)
    c = torch.zeros(hidden_size)

    outputs = []

    for x_t in sequence:

        combined = torch.cat([x_t, h])

        gates = combined @ W + b

        i, f, o, g = torch.chunk(gates, 4)

        i = torch.sigmoid(i)
        f = torch.sigmoid(f)
        o = torch.sigmoid(o)
        g = torch.tanh(g)

        c = f * c + i * g
        h = o * torch.tanh(c)

        outputs.append(h)

    return torch.stack(outputs)


# Forward direction
forward_output = lstm_forward(x)

# Backward direction
backward_output = lstm_forward(torch.flip(x, dims=[0]))
backward_output = torch.flip(backward_output, dims=[0])

# Combine both directions
output = torch.cat(
    [forward_output, backward_output],
    dim=1
)

print("BiLSTM Output Shape:", output.shape)
```
