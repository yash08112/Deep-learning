**Task 12** — Convolutional Autoencoder for Anomaly Detection

```python
import torch
import torch.nn as nn


class Autoencoder(nn.Module):

    def __init__(self):

        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(
                1, 16, 3,
                stride=2,
                padding=1
            ),
            nn.ReLU(),

            nn.Conv2d(
                16, 32, 3,
                stride=2,
                padding=1
            ),
            nn.ReLU()
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(
                32, 16, 4,
                stride=2,
                padding=1
            ),
            nn.ReLU(),

            nn.ConvTranspose2d(
                16, 1, 4,
                stride=2,
                padding=1
            ),
            nn.Sigmoid()
        )

    def forward(self, x):

        encoded = self.encoder(x)

        decoded = self.decoder(encoded)

        return decoded


model = Autoencoder()

# Normal image
x = torch.rand(4, 1, 64, 64)
output = model(x)
# Reconstruction error
error = torch.mean(
    (x - output) ** 2,
    dim=(1, 2, 3)
)

print("Reconstruction Error:")
print(error)

# Detect anomalies
threshold = 0.1

for value in error:

    if value > threshold:
        print("Anomaly detected")
    else:
        print("Normal image")
```
