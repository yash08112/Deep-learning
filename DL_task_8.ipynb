**Task 8** — U-Net Segmentation with Dice Loss

```python
import torch
import torch.nn as nn


# -------------------------
# Dice Loss
# -------------------------

def dice_loss(pred, target):

    smooth = 1e-6

    pred = torch.sigmoid(pred)

    intersection = (pred * target).sum()

    dice = (
        2 * intersection + smooth
    ) / (
        pred.sum() + target.sum() + smooth
    )

    return 1 - dice


# -------------------------
# Simple U-Net
# -------------------------

class SimpleUNet(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv2d(1, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU()
        )

        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(
                32, 16, 2, stride=2
            ),
            nn.ReLU(),

            nn.Conv2d(
                16, 1, 3, padding=1
            )
        )

    def forward(self, x):

        x = self.encoder(x)
        x = self.decoder(x)

        return x


model = SimpleUNet()

image = torch.randn(2, 1, 64, 64)
mask = torch.randint(
    0, 2,
    (2, 1, 64, 64)
).float()

output = model(image)

loss_dice = dice_loss(output, mask)

bce = nn.BCEWithLogitsLoss()
loss_bce = bce(output, mask)

loss = loss_dice + loss_bce

print("Dice Loss:", loss_dice.item())
print("BCE Loss:", loss_bce.item())
print("Total Loss:", loss.item())
```
